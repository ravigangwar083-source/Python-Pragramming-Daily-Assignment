```python
#Tuple datatype(  )
data=(10,20,30,40,50) ##type 1 to create tuple
print(type(data))
data2=10,20,30,40,50 ##type 2 to create tuple
print(type(data2))
```

```python
data3=(10) ##if you use parenthesis in single element to create tuple then it will print integer type.
print(type(data3))
data4=(10,) ##type 3 to create tuple
print(type(data4))
```

```python
data=(10,20,30,40,50)
print(data[0:3]) ##slicing
print(data[1]) ##indexing
print(data[-1]) ##negative indexing
```

```python
data=(10,20,30,40,50,30,20) ##duplicate Elements
print(data)
```

```python
#Tuple is Immutable
data=(10,20,30,40,50)
data[0]=88 ##TypeError: 'tuple' object does not support item assignment 
```

```python
#Tuple Packing
studentdata="Rahul","MCA AI",84.33 
print(studentdata)
print(type(studentdata))
```

```python
#Tuple Unpacking
studentdata="Rahul","MCA AI",84.33 
Name,Course,AvgMarks=studentdata
print(Name)
print(Course)
print(AvgMarks)
n,*ab=studentdata ##Here n will return first element and *ab will return rest of element in tuple form
print(n)
print(ab)
```

```python
studentdata="Rahul","MCA AI",84.33,22
name,*ab,age=studentdata
print(name)
print(ab)
print(age)
Name,Course=studentdata #ValueError: too many values to unpack (expected 2, got 3)
```

```python
studentdata="Rahul","MCA AI",84.33,23.44 
name,*ab,age=studentdata
print(name)
print(ab)
print(age)
```

```python
#Enumerate Method
data=("Apple","Banana","Papaya","Guava")
for index, value in enumerate(data): ## Will return from inxing 0.
    print(index, value)

for index, value in enumerate(data, start=1): ## If want to return from indexing 1.
    print(index, value)
```

```python
#comparison Operator
x=10
y=20
print(x==y)
```

```python
#Logical operator
#and
x=10
y=20
print(x<y and x>y)
print(x<y and x==y)
print(x>y and x<y)
print(x>y and x==y)
```

```python
#or Operator
x=10
y=20
print(x<y or x>y)
print(x<y or x==y)
print(x>y or x<y)
print(x==y or y<x)
```

```python
# Membership operator
data=[10,20,30,40,50]
print(10 in data)
print(10 not in data)
print(100 not in data)
```

```python
#Identity operators
num1=10
num2=10
print(num1==num2)
#print(num1 is num2)
print(id(num1))
print(id(num2))
```
