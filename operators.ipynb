```python
x=7
y=2
##Floor Division
print(x//y) ## after divide nearest value (3 and 4) will return smallest i.e 3
print(-x//y) ## after divide narest value (-3 and -4) will return smallest i.e -4.
##Exponentiation
print(x**y) ## Returns the power value. i.e x raises to the power y.
##Modulus
print(x%y) ## Returns the remainder.
```
