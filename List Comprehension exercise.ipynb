```python
employees=[
    {"name": "Amit", "salary": 45000},
    {"name": "Rahul", "salary": 60000},
    {"name": "Neha", "salary": 75000},
    {"name": "Rohit", "salary": 35000}]
result=[(employee["name"],employee["salary"]) for employee in employees if employee["salary"]>50000]
print(result)
```

```python
employees=[
    {"name": "Amit", "department": "IT"},
    {"name": "Rahul", "department": "HR"},
    {"name": "Neha", "department": "IT"},
    {"name": "Rohit", "department": "Sales"}]
newdata=[i["name"] for i in employees if i["department"]=="IT"]
print(newdata)
```

```python
employees=[
    {"name": "Amit", "department": "IT", "salary": 55000},
    {"name": "Rahul", "department": "HR", "salary": 70000},
    {"name": "Neha", "department": "IT", "salary": 80000},
    {"name": "Rohit", "department": "IT", "salary": 45000}]
newdata=[(i["name"],i["department"], i["salary"])for i in employees if i["salary"]>60000 and i["department"]=="IT"]
print(newdata)
```
