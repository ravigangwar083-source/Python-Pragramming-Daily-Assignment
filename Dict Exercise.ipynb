```python
students={"Rahul": 75, "Amit": 32, "Priya": 88, "Neha": 39, "Rohit":55}
for name,marks in students.items():
    if marks>=40:
        if marks>=80:
            print(name,marks,"Excellent")
        else:
            if marks>=60:
                print(name,marks,"Good")
            else:
                if marks>=40:
                    print(name,marks,"Average")
    else:
        print(name,marks,"Fail")
```

```python
product={"Laptop":10, "Mouse":3, "Keyboard":0, "Monitor":7, "Printer":2}
for name,stock in product.items():
    if stock==0:
        print(name,"Out of Stock")
    else:
        if stock<=5:
            print(name,"Low Stock")
        else:
            if stock>5:
                print(name,"Available Stock")
```

```python
#Employee Salary and Department
employees = {
    "Rahul": {"salary": 60000, "department": "IT"},
    "Amit": {"salary": 40000, "department": "HR"},
    "Priya": {"salary": 70000, "department": "IT"},
    "Neha": {"salary": 45000, "department": "Finance"}
}
for name, details in employees.items():
    salary = details["salary"]
    dept = details["department"]
    if salary >= 50000:
        if dept == "IT":
            print(name, "Senior IT Employee")
        else:
            if dept == "HR":
                print(name, "Senior HR Employee")
            else:
                print(name, "Senior Employee")
    else:
        print(name, "Junior Employee")
```

```python
#Check Even/Odd Values
data={
    "a": 10,
    "b": 15,
    "c": 25,
    "d": 30,
    "e": 40
}
for key, value in data.items():
    if value % 2 == 0:
        if value > 20:
            print(key, "Large Even")
        else:
            print(key, "Even")
    else:
        print(key, "Odd")
```

```python
#Shopping Cart Classification
products = {
    "Mobile": 15000,
    "Mouse": 500,
    "Laptop": 60000,
    "Keyboard": 1200
}
for product, price in products.items():
    if price >= 1000:
        if price >= 5000:
            print(product, "Premium Product")
        else:
            print(product, "Expensive Product")
    else:
        print(product, "Affordable Product")
```

```python
#Student Eligibility
students = {
    "Rahul": {"attendance": 85, "marks": 70},
    "Amit": {"attendance": 60, "marks": 80},
    "Priya": {"attendance": 90, "marks": 35},
    "Neha": {"attendance": 78, "marks": 55}
}
for name, details in students.items():
    attendance = details["attendance"]
    marks = details["marks"]
    if attendance >= 75:
        if marks >= 40:
            print(name, "Eligible")
        else:
            print(name, "Failed in Exam")
    else:
        print(name, "Not Eligible")
```

```python
#Bank Account Classification
accounts = {
    "Rahul": 25000,
    "Amit": 5000,
    "Priya": 0,
    "Neha": -2000
}
for name, balance in accounts.items():
    if balance > 0:
        if balance >= 10000:
            print(name, "Premium Account")
        else:
            print(name, "Regular Account")
    else:
        if balance == 0:
            print(name, "No Balance")
        else:
            print(name, "Overdraft")
```

```python
#Employee Experience (Nested Dictionary)
employees = {
    "Rahul": {"salary": 60000, "experience": 6},
    "Amit": {"salary": 55000, "experience": 3},
    "Priya": {"salary": 45000, "experience": 4},
    "Neha": {"salary": 35000, "experience": 1}
}
for name, details in employees.items():
    salary = details["salary"]
    experience = details["experience"]
    if salary >= 50000:
        if experience >= 5:
            print(name, "Senior")
        else:
            print(name, "Mid-Level")
    else:
        if experience >= 3:
            print(name, "Experienced Junior")
        else:
            print(name, "Fresher/Junior")
```

```python
#Student Marks & Attendance Classification
students = {
    "Rahul": {"marks": 85, "attendance": 90},
    "Amit": {"marks": 35, "attendance": 80},
    "Priya": {"marks": 65, "attendance": 70},
    "Neha": {"marks": 45, "attendance": 85},
    "Rohit": {"marks": 90, "attendance": 60}
}
for name, details in students.items():
    marks = details["marks"]
    attendance = details["attendance"]
    if attendance >= 75:
        if marks >= 80:
            print(name, "Distinction")
        else:
            if marks >= 60:
                print(name, "First Division")
            else:
                if marks >= 40:
                    print(name, "Pass")
                else:
                    print(name, "Fail")
    else:
        print(name, "Not Eligible")
```

```python
#Square 
n=4
for i in range (n):
    for j in range (n):
        print("*",end=" ")
    print()
```

```python
#User Based Square
n=int(input("Enter the rows:"))
m=int(input("Enter the column:"))
for i in range (n):
    for j in range (m):
        print("*",end=" ")
    print()
```

```python
#User Based Triangle
n=int(input("Enter the rows:"))
for i in range (1,n):
    for j in range (i):
        print("*",end=" ")
    print()
```

```python
#Reverse Triangle (User Based)
n=int(input("Enter the rows:"))
for i in range (n,0,-1):
    for j in range (i):
        print("*",end=" ")
    print()
```
