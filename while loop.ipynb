```python
#Exercise 4: Find the sum of numbers from 1 to 10
i=1
sum=0
while i <= 10:
    sum+=i
    i+=1
print("Sum =", sum)
```

```python
#Exercise 5: Find the sum of even numbers from 1 to 100
i = 1
sum = 0
while i <= 100:
    if i % 2 == 0:
        sum = sum + i
    i += 1
print("Sum of even numbers =", sum)
```

```python
#Exercise 8: Take a number from user and print its multiplication table
num = int(input("Enter a number: "))
i = 1
while i <= 10:
    print(num, "x", i, "=", num * i)
    i+=1
```

```python
#Exercise 9: Print numbers from 1 to 100 divisible by both 3 and 5
i = 1
while i <= 100:
    if i%3==0 and i%5==0:
        print(i)
    i+=1
```

```python
#Exercise 10: Count even and odd numbers between 1 and 50
i = 1
even = 0
odd = 0
while i <= 50:
    if i % 2 == 0:
        even += 1
    else:
        odd += 1
    i += 1
print("Even numbers =", even)
print("Odd numbers =", odd)
```
