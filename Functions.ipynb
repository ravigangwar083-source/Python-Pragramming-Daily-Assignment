```python
x=30  # Global Variable
def show():
    global x
    print("Global Variable: ",x)  # 30
show()
print("Global Variable: ",x)  # 30

def display():
    global x
    print("Global Variable: ",x)  # 30
display()
```

```python
balance=3000  # Global Variable
def show():
    global balance
    deposit=1000
    print("Deposit Balance: ",deposit)

    balance=balance+deposit  # 4000

show()
print("Current Balance After Deposit: ",balance)
```

```python
balance=3000  # Global Variable
def show():   # Function without parameter
    deposit=1000
    print("Deposit Balance: ",deposit)
    print("Initial Balance was : ", globals() ["balance"])   # One time use

show()
print("Current Balance After Deposit: ",balance)
```

```python

# args
def add(*args):  # tuple
    total=0  # 30
    for ex in args:
        total=total+ex  # 150
    print("Sum : ",total)
    
add(10,20)   # args=(10,20)
add(40,50,60)  # args=(40,50,60)
add(20,30,40,50)  # args=(20,30,40,50)
add(20,30,40,50,60)  # args=(20,30,40,50,60)
```

```python
def add(*args):
    print("args[0]",args[0])  # indexing
    print("args[1]",args[1])  # indexing
    print("args[2]",args[2])  # indexing
    print("args[3]",args[3])  # indexing
    print("args[4]",args[4])  # indexing
    print(args)   # (20,30,40,50,60) Tuple
                  # [20,30,40,50,60] List

add(20,30,40,50,60)   # args=((20,30,40,50,60)
```

```python
def Employeedetails(**kwargs):
    for key, value in kwargs.items():
        print(key,":", value)

    print("===============================")
Employeedetails(name="Mohit",Role="Developer",Department="IT")
Employeedetails(name="Mohit",Department="IT")
Employeedetails(name="Mohit")
```

```python
def calculate_avg(*marks):
    if not marks:
        return "Error: No Marks Provided"
    for ex in marks:
        if not isinstance(ex,(int,float)) or isinstance(ex, bool):
            return "Error: Marks Must be Numeric"
        if ex<0 or ex>100:
            return "Error: Marks Must be between 0 and 100"
        return sum(marks)/len(marks)

print(calculate_avg(55,45,58,97,0,87,67))
print(calculate_avg())
print(calculate_avg(45,76,True))
```

```python
def even_number(start,end):
    for num in range(start,end+1):
        if(num%2==0):
            yield num  #2

ex=even_number(1,10)   # 2, 4, 6, 8, 10
print(next(ex,"Error: iteration Over"))
print(next(ex,"Error: iteration Over"))
print(next(ex,"Error: iteration Over"))
print(next(ex,"Error: iteration Over"))
print(next(ex,"Error: iteration Over"))
print(next(ex,"Error: iteration Over"))   # Error: iteration Over
```

```python
# Student Records

students = [
    {"name": "Aman", "marks": [80,75,90]},
    {"name": "Riya", "marks": [35,40,30]},
    {"name": "Rahul", "marks": [90,85,95]},
    {"name": "Neha", "marks": [60,70,65]}
]

def generate_result(student_record):
    for student in student_record:
        name=student['name']
        marks=student['marks']

        avg=sum(marks)/len(marks)
        if avg>=40:
            status="Pass"
        else:
            status="Fail"

        yield {
            "name":name,
            "average":avg,
            "status":status
        }

ex=generate_result(students)
print(next(ex))
print(next(ex))
print(next(ex))
print(next(ex))
```

```python

# Student Records

students = [
    {"name": "Aman", "marks": [80,75,90]},
    {"name": "Riya", "marks": [35,40,30]},
    {"name": "Rahul", "marks": [90,85,95]},
    {"name": "Neha", "marks": [60,70,65]}
]

def generate_result(student_record):
    for student in student_record:
        name=student['name']
        marks=student['marks']

        avg=sum(marks)/len(marks)
        if avg>=40:
            status="Pass"
        else:
            status="Fail"

        yield {
            "name":name,
            "average":avg,
            "status":status
        }

for result in generate_result(students):
    print(result['name'])
    print(result['average'])
    print(result['status'])
    print("-"*25)
```
