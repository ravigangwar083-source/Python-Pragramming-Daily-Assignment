```python
n=6
for i in range(1,n+1): #5 no of rows
    for j in range(0,n-i): #(n-1)=4 >>>> 4 times
        print("  ",end="")
    for j in range (0,i): # first time this loop will 1 time
        print("* ",end="")
    print()
```

```python
n=int(input("Enter Number of rows:"))
for i in range(1,n+1):
    for j in range(n-i): # 4 times, 3 times, 2 times, 1 time, 0 times
        print(" ",end=" ")
    for j in range(0,(2*i)-1): #2*1-1
        print("*",end=" ")
    print()
```
