```python
#methods of list
studentname=["Ravi","pathak","Dhruv","Ravi","Aman","Anurag"]
print(len(studentname))

#append
studentname.append("Saurabh") #add one element at the end
print(studentname)
studentname.append(["Rahul","Priya"])
print(studentname)

#insert
studentname.insert(1,"Shukla") #insert element at given index
print(studentname)

#Extend
studentname.extend(["Rahul","Rohit"])
print(studentname)

#remove
studentname.remove("Ravi") #Will remove first matching value
print(studentname)

studentname.remove("sumit") #ValueError: list.remove(x): x not in list

```

```python
#pop
#studentname.remove("sumit") #ValueError: list.remove(x): x not in list
print(studentname.pop()) #delete last element
print(studentname.pop(1)) #delete value of index 1
print(studentname)
print(studentname.pop(20)) #IndexError: pop index out of range
```

```python
#shallow copy
a=[20,10,40,30,70]
b=a.copy()
print(b)
b[3]=60
print(a)
print(b)
print(id(a)) # In shallow copy id change in normal list. 
print(id(b))
```

```python
#shallow copy (nested list)
data=[[12,13,16]
      ,[22,27,45]
     ]
print(data)
newdata=data.copy()  ## In nested list object id do not changes.
print(newdata)
newdata[1][2]=400
print(data)
print(newdata)
print(newdata[-1])
print(id(data[0]))
print(id(data[1]))
print(id(newdata[0]))
print(id(newdata[1]))
```

```python
##deep copy 
data=[[12,13,16]
      ,[22,27,45]
     ]
print(data)
newdata=copy.deepcopy(data) #In this id of all copied objects/elements will be different
newdata[0][2]=500
print(newdata)
print(id(data[0]))
print(id(data[1]))
print(id(newdata[0]))
print(id(newdata[1]))
```
