```python

class Employee: # E_ID, E_Name,E_Salary, E_Company
    def addEmpDetails(self, E_ID, E_Name, E_Salary, E_Company): 
        self.id=E_ID
        self.name=E_Name
        self.salary=E_Salary
        self.company=E_Company

    def showEmpDetails(self):
        print("Employee ID = ",self.id)
        print("Employee Name = ",self.name)
        print("Employee Salary = ",self.salary)
        print("Employee Company = ",self.company)

e1=Employee()
e1.addEmpDetails(111,"Raj",12000.00,"HCL")
e1.showEmpDetails()

print("="*25)

e2=Employee()
e2.addEmpDetails(112,"Riddhima",70000.00,"HCL")
e2.showEmpDetails()

print("="*25)
```

```python

# Print details in one line :->

class Employee: # E_ID, E_Name,E_Salary, E_Company
    def addEmpDetails(self, E_ID, E_Name, E_Salary, E_Company): 
        self.id=E_ID  # >>> E_ID is a Local Variable bur self.id ia s instance variable
        self.name=E_Name
        self.salary=E_Salary
        self.company=E_Company

    def showEmpDetails(self):
        print(self.id, end=" ")
        print(self.name, end=" ")
        print(self.salary, end=" ")
        print(self.company)

e1=Employee()
e1.addEmpDetails(111,"Raj",12000.00,"HCL")
e1.showEmpDetails()

print("="*30)
```

```python

# OOP (CONSTRUCTOR) :->>>

class Employee: # E_ID, E_Name,E_Salary, E_Company
    def __init__(self, E_ID, E_Name, E_Salary, E_Company): 
        self.id=E_ID
        self.name=E_Name
        self.salary=E_Salary
        self.company=E_Company

    def showEmpDetails(self):
        print("Employee ID = ",self.id)
        print("Employee Name = ",self.name)
        print("Employee Salary = ",self.salary)
        print("Employee Company = ",self.company)

e1.addEmpDetails(111,"Raj",12000.00,"HCL")  # __new__()   __init__()
e1.showEmpDetails()

print("="*25)

e2.addEmpDetails(112,"Riddhima",70000.00,"HCL")
e2.showEmpDetails()

print("="*25)
```

```python


class Employee: 

    E_Company="HCL"    # STATIC VARIABLE / OR CLASS VARIABLE 
    
    def __init__(self, E_ID, E_Name, E_Salary): # E_ID ->> LOCAL VARIABLE
        self.id=E_ID   # INSTANCE VARIABLE -->> self.id=E_ID
        self.name=E_Name
        self.salary=E_Salary

    def showEmpDetails(self):
        print("Employee ID = ",self.id)
        print("Employee Name = ",self.name)
        print("Employee Salary = ",self.salary)
        print("Employee Company = ",Employee.E_Company)

e1=Employee(111,"Raj",12000.00)  # __new__()   __init__()
e1.showEmpDetails()

print("="*25)

e2=Employee(112,"Riddhima",70000.00)
e2.showEmpDetails()

print("="*25)
```

```python

# BANK ACCOUNT DETAILS -->>>

class Bank: 
    def __init__(self, AccountNumber, AccountHolderName, Amount):   # def __init__(self, AccountNumber=0, AccountHolderName="", Amount=0): 
        self.AccountNumber=AccountNumber
        self.AccountHolderName=AccountHolderName
        self.Amount=Amount

    def showAccountDetails(self):
        print("Account Number = ",self.AccountNumber)
        print("Account Name = ",self.AccountHolderName)
        print("Account Amount = ", self.Amount)

e1=Bank(111,"Raj",12000.00)  # __new__()   __init__()
e1.showAccountDetails()

print("="*25)

e2=Bank(112,"Riddhima",70000.00)
e2.showAccountDetails()

print("="*25)
```
